```python
from sqlalchemy import create_engine
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import Column, String, Integer
from sqlalchemy.orm import sessionmaker
```

```python
engine = create_engine('mysql+pymysql://root:1316@localhost:3306/cinema')
Base = declarative_base()
Session = sessionmaker(bind=engine)
session = Session()
```

Entidades

```python
class Filmes(Base):
    __tablename__ = "filmes"

    titulo = Column(String, primary_key=True)
    genero = Column(String, nullable=False)
    ano = Column(Integer, nullable=False)

    def __repr__(self):
        return f"filme [titulo={self.titulo}, genero={self.genero}, ano={self.ano}]"
```

SQL

```python
data_insert = Filmes(titulo="Gato de Botas", genero="Acao", ano=2001)
session.add(data_insert)
session.commit()
```

```python
session.query(Filmes).filter(Filmes.titulo=="Dracula").delete()
session.commit()
```

```python
session.query(Filmes).filter(Filmes.genero=="Drama").update({"ano": 2000})
session.commit()
```

```python
data =session.query(Filmes).all()
print(data)
print(data[0].titulo)
```

```python
session.close()
```
